# Candidate terpene identity census

Audit the frozen aggregate union report from the upstream census. This is a report validation, not a reconstruction from molecule-level records.

```python
import json
from pathlib import Path

report = json.loads(Path("data/reports/coconut-terokit-union-20260903.json").read_text())
report
```

```python
sources = report["sources"]
left = sources["coconut_terpenoids"]["unique_identities"]
right = sources["terokit_classified"]["unique_identities"]
overlap = report["cross_source_overlap"]
union = report["cross_source_union"]
assert left + right - overlap == union
print({"coconut_identities": left, "terokit_identities": right, "overlap": overlap, "union": union})
```
